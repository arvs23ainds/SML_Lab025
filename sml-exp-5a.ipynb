```python
import numpy as np
from scipy import stats
```

```python
sample_mean = 8
sample_std = 2
n=20
mu_0 = 0
```

```python
t_stats = (sample_mean-mu_0) / (sample_std/np.sqrt(n))
```

```python
df=n-1
```

```python
p_value = 2*(1-stats.t.cdf(abs(t_stats),df=df))
```

```python
print("t-statistics:", t_stats)
print("Degree of Freedom:", df)
print("p-value:", p_value)
```

```python
alpha = 0.5
if p_value < alpha:
    print("Reject H0 : the drug significantly affects the heart rate")
else:
    print("Fail to Reject H0 : No Significant effect Detected")
```
